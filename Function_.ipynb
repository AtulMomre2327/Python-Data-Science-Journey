Example_1: Extract a list of fisrt letters from a sentence

```python
heisenberg_quote = "It ceases to exist without me. no,you clearly don't know who you're"
heisenberg_quote =  heisenberg_quote.split()
print(heisenberg_quote)
for i in heisenberg_quote:
  print(i[0 ])
```

```python
def extract_letter(heisenberg_quote):
  heisenberg_quote =  heisenberg_quote.split()
  print(heisenberg_quote)
  k = []
  for i in heisenberg_quote:
    k.append(i[0])
```

```python
a = 'i am good'
extract_letter(a)
```
