```python
lst = [1,2,3,4,5,6,7,8]
even_sum = 0
odd_sum = 0
for i in lst:
    if(i%2==0):
        even_sum = even_sum+i
    else:
        odd_sum = odd_sum+i
print("even sum is {}".format(even_sum))
print("odd sum is {}".format(odd_sum))
```
