```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
#
```

```python
df = pd.read_csv('/content/drive/MyDrive/Data  Set/adult.csv')
```

```python
df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df['income'].value_counts()
```

```python
df = pd.get_dummies(
    df,
    columns=[
        'workclass',
        'education',
        'marital.status',
        'occupation',
        'relationship',
        'race',
        'sex',
        'native.country'
    ],
    drop_first=True
)
```

```python
df['income'] = df['income'].map({'<=50K':0,'>50K':1})
```

```python
df['income'].value_counts()
```

```python
df.columns
```

```python
x = df.drop('income', axis=1)
y = df['income']
```

```python
x.head()
```

```python
y.head()
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size = 0.2, random_state = 42)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
model = LogisticRegression()
```

```python
model.fit(x_train,y_train)
```

```python
train_pred = model.predict(x_train)
test_pred = model.predict(x_test)
```

```python
from  sklearn.metrics import accuracy_score, confusion_matrix, classification_report,precision_score,recall_score,f1_score
```

```python
print('Train Confusion Matrix :-\n',confusion_matrix(y_train,train_pred))
print('Train Accuracy :-',accuracy_score(y_train,train_pred))
print('Train Precision :-',precision_score(y_train,train_pred))
print('Train Recall :-',recall_score(y_train,train_pred))
```

```python
print('Test Confusion Matrix :-\n',confusion_matrix(y_test,test_pred))
print('Test Accuracy :-',accuracy_score(y_test,test_pred))
print('Test Precision :-',precision_score(y_test,test_pred))
print('Test Recall :-',recall_score(y_test,test_pred))
```
