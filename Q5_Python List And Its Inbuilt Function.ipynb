# Real Ex List Ke Aur Methods Samjhate Hain! 😊

```python
#1️⃣ append() — End Mein Add Karo
my_list = [1, 2, 3]
my_list.append(4)
print(my_list)  # [1, 2, 3, 4]

# Matlab: List ke end mein ek cheez daalo
```

```python
# 2️⃣ insert() — Kahi Bhi Add Karo
my_list = [1, 2, 3]
my_list.insert(1, 99)
print(my_list)  # [1, 99, 2, 3]

# Matlab: Jis jagah chahiye wahan daalo
#insert(1, 99) → position 1 par 99 daalo
```

```python
#3️⃣ remove() — Element Hatao
my_list = [1, 2, 3, 2]
my_list.remove(2)
print(my_list)  # [1, 3, 2]

# Matlab: Pehla wala element hatao

```

```python
#4️⃣ pop() — Last Element Nikalo
my_list = [1, 2, 3]
my_list.pop()
print(my_list)  # [1, 2]

# Ya koi bhi position se
my_list.pop(0)
print(my_list)  # [2]

# Matlab: Element nikalo aur hatao bhi
```

```python
#5️⃣ sort() — Chhote Se Bade Karo
my_list = [3, 1, 4, 1, 2]
my_list.sort()
print(my_list)  # [1, 1, 2, 3, 4]

# Ulta karna ho toh
my_list.sort(reverse=True)
print(my_list)  # [4, 3, 2, 1, 1]

#Matlab: List ko order mein lagao

```

```python
#6️⃣ reverse() — Ulta Karo
my_list = [1, 2, 3, 4]
my_list.reverse()
print(my_list)  # [4, 3, 2, 1]

# Matlab: Poori list ko ulta kar do
```

```python
# 7️⃣ len() — Kitne Elements Hain
my_list = [1, 2, 3, 4, 5]
print(len(my_list))  # 5

# Matlab: List mein total kitni cheezein hain
```

```python
#8️⃣ index() — Kahan Hai Element
my_list = [10, 20, 30, 40]
print(my_list.index(30))  # 2

#Matlab: Element kis position par hai

```

```python
#9️⃣ count() — Kitni Baar Aaya
my_list = [1, 2, 2, 3, 2]
print(my_list.count(2))  # 3

#Matlab: Element kitni baar aaya — yeh tumhe pata hai! 😄
```

```python
#🔟 clear() — Sab Hatao
my_list = [1, 2, 3]
my_list.clear()
print(my_list)  # []

#Matlab: Poori list khali kar do
```

```python
#1️⃣1️⃣ copy() — Copy Banao
my_list = [1, 2, 3]
new_list = my_list.copy()
print(new_list)  # [1, 2, 3]
#Matlab: List ki ek copy banao
```

```python
#1️⃣2️⃣ extend() — Doosri List Jodo
list1 = [1, 2, 3]
list2 = [4, 5, 6]
list1.extend(list2)
print(list1)  # [1, 2, 3, 4, 5, 6]

#Matlab: Ek list ko doosri list se jodo
```
