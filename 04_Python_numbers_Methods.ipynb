### Ex math.exp()

```python
import math

# Formula: Amount = Principal * e^(rate * time)
principal = 100000  # ₹1 Lakh invest kiye
rate = 0.08         # 8% interest rate (0.08)
time = 5            # 5 saal ke liye

final_amount = principal * math.exp(rate * time)
print(f"5 saal baad ₹{final_amount:.2f} milenge.")
# Output: 5 saal baad ₹149182.47 milenge.
 
```

```python
import math

bill = 455.75
paise, rupees = math.modf(bill)

# Paise float me aate hain, unhe round ya multiply kiya jata hai
paise = round(paise * 100) 
rupees = int(rupees)

print(f"Aapka bill {rupees} Rupaye aur {paise} Paise hai.")
# Output: Aapka bill 455 Rupaye aur 75 Paise hai.
```
